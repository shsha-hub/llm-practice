# 🟡 미션 2 — 내가 고른 글을 요약한다

수업 `## 9` 에서 한 것을 **내가 고른 글 2개**로 한다 — 짧은 글 하나, 긴 글 하나.

## 낼 것

1. 두 글의 **요약 결과 캡처**
2. 한 줄 — 요약이 원문에서 **무엇을 남기고 무엇을 버렸나**

```python
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

kobart_tokenizer = AutoTokenizer.from_pretrained("gogamza/kobart-base-v2")
kobart = AutoModelForSeq2SeqLM.from_pretrained("gogamza/kobart-summarization")
kobart.eval()
```

## 1. 글을 가져온다

✍️ **아래 두 글을 내가 고른 글로 바꾼다.** 뉴스, 블로그, 학교 공지 — 무엇이든 좋다.
짧은 글은 3~4문장, 긴 글은 한두 문단 정도.

```python
짧은글 = (
    "우리나라 폭염에 변수가 될 수 있는 강력한 태풍이 서태평양에서 북상하고 있습니다."
    "북서쪽으로 이동해 8월 4일(화)에는 일본 도쿄 남쪽 약 1,220㎞ 해상에 태풍이 위치하겠습니다."
)  # ✍️ 바꾼다

긴글 = (
    "카카오뱅크 노동조합이 성과 보상체계 개편을 요구하며 인터넷전문은행 출범 이후 첫 파업에 들어갔다."
    "31일 민주노총 화섬식품노조 카카오지회에 따르면 이날 카카오뱅크 법인 조합원 1000여명 중 700명 이상이 여나를 사용해 파업에 참여했다."
    "별도의 장외 집회 없이 업무에 빠지는 준법 투쟁 성격의 파업이다."
    "이번 차업은 카카오뱅크 출범 이후 최초이자 국내 인터넷전문은행 업계 전체를 통틀어 처음 있는 일이다."
    "양측은 올해 임금 교섭 솨정에서 성과급 지급과 5~6%수준의 연봉 인상안을 두고 첨예하게 맞섰으나 합의점을 차지 못했다."
    "다만 오프라인 영업점이 없는 인터넷은행의 특성과 여름휴가 성수기가 맞물리면서 고객들이 체감하는 서비스 영향은 크지 않을 것으로 보인다."
    "현재까지 관련 고객 불만 신고 역시 접수되지 않았다. 사측은 핵심 업무 인력 등을 중심으로 업무연속성계획을 가동해 서비스 차질을 막고 있다."
    "카카오뱅크 관계자는 “회사는 대화를 통해 원만한 해결 방안을 찾기 위해 지속해서 노력하겠다”고 밝혔다. 노조 측 역시 향후 투쟁 방식과 일정을 두고 내부 논의를 이어가고 있어 향후 양측의 교섭 재개 여부에 이목이 집중된다."
)  # ✍️ 바꾼다
```

## 2. 요약한다 — 수업 `## 9` 그대로

⚠️ 시작/끝 표시를 직접 붙이는 것을 잊지 말 것. 빼먹으면 요약이 엉뚱해진다.

```python
for text in [짧은글, 긴글]:
    ids = kobart_tokenizer.encode(text)
    input_ids = torch.tensor(
        [[kobart_tokenizer.bos_token_id] + ids + [kobart_tokenizer.eos_token_id]]
    )
    summary_ids = kobart.generate(input_ids, max_length=64)   # 최대 64토큰에서 멈춤
    summary = kobart_tokenizer.decode(summary_ids[0], skip_special_tokens=True)  # 기호 빼고 글자만

    print(f"원문 ({input_ids.shape[1]}토큰): {text[:50]} …")
    print(f"요약: {summary}")
    print()
```
